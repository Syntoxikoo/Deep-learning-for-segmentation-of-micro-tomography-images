````python
from torch.utils.data import Dataset, DataLoader
from PIL import Image
import torchvision.transforms as T
import os, torch, numpy as np, matplotlib.pyplot as plt, tifffile as tiff
import torch.nn.functional as F
import sys
import os
````

````python

class MicroCTDatasetTest(Dataset):
    def __init__(self, img_dir, mask_dir=None):
        self.img_dir = img_dir
        self.mask_dir = mask_dir
        self.img_files = sorted([f for f in os.listdir(img_dir) if f.lower().endswith(".tif")])
        if mask_dir:
            self.mask_files = [f.replace("image_v2_", "image_v2_mask_") for f in self.img_files]
        else:
            self.mask_files = None
        
        self.resize = T.Resize((768,768), interpolation=T.InterpolationMode.BILINEAR)

    def __len__(self): 
        return len(self.img_files)

    def __getitem__(self, idx):
        img = tiff.imread(os.path.join(self.img_dir, self.img_files[idx]))
        if img.ndim == 3: img = img[img.shape[0]//2]

        img = img.astype(np.float32)
        img /= img.max()

        img = Image.fromarray(img)
        img = self.resize(img)
        img = np.array(img, dtype=np.float32)
        img = torch.tensor(img).unsqueeze(0)

        if self.mask_dir is None:
            return img, None

        mask = tiff.imread(os.path.join(self.mask_dir, self.mask_files[idx]))
        if mask.ndim == 3: mask = mask[mask.shape[0]//2]
        mask = (mask > 0).astype(np.float32)
        mask = Image.fromarray(mask)
        mask = self.resize(mask)
        mask = torch.tensor(np.array(mask)).unsqueeze(0)

        return img, mask
````

````python
# Add the parent directory of 'src' to sys.path
sys.path.append(os.path.abspath(os.path.join('..', )))

# Now you can import
from utils.dataset_loader_vit import TOMODatasetViT

test_ds = TOMODatasetViT(
    img_dir="../../datas/original/test/imgs",
    label_dir="../../datas/original/test/labels",
    split=None,
    resized_shape=[768, 768],
    transform=None
)

test_loader = DataLoader(test_ds, batch_size=1, shuffle=False)
print("Test samples:", len(test_ds))
````

````python
def dice_metric(p, t, eps=1e-6):
    p = (p.reshape(-1) > 0.5).float()
    t = t.reshape(-1).float()
    i = (p * t).sum()
    u = p.sum() + t.sum()
    return (2*i + eps) / (u + eps)

def dice_metric_inv(p, t, eps=1e-6):
    # invert both masks: 1 → 0, 0 → 1
    p = (p.reshape(-1) > 0.5).float()
    t = t.reshape(-1).float()
    p = 1 - p
    t = 1 - t
    i = (p * t).sum()
    u = p.sum() + t.sum()
    return (2*i + eps) / (u + eps)
````

````python
# Load models
from models.Unet_ViT import U_net_ViT
from models.Unet import U_net
````

````python
 # Load U-Net

device=torch.device("cuda" if torch.cuda.is_available() else "cpu")

model_unet=U_net(
    encode_in=(1,64,128,256), encode_out=(64,128,256,512),
    decode_in=(1024,512,256,128), decode_out=(512,256,128,64),
    normalize=True
)
model_unet.load_state_dict(torch.load("../../models/predicted_models/unet_trained.pth", map_location=device))
model_unet.to(device); model_unet.eval()
````

````python
# Load U-Net ViT

device=torch.device("cuda" if torch.cuda.is_available() else "cpu")

model_vit=U_net_ViT(
    encode_in=(1,64,128,256), encode_out=(64,128,256,512),
    decode_in=(1024,512,256,128), decode_out=(512,256,128,64),
    normalize=True
)
model_vit.load_state_dict(torch.load("../../models/predicted_models/unet_vit_trained.pth", map_location=device))
model_vit.to(device); model_vit.eval()
````

````python
# Eval U-Net
all_d = []
model_unet.eval()

with torch.no_grad():
    for i, (img, gt) in enumerate(test_loader):
        img = img.to(device)
        gt = gt.to(device)

        out = model_unet(img)
        if isinstance(out, tuple):
            out = out[0]

        pred = torch.sigmoid(out)
        pred_bin = (pred > 0.5).float()
    
        d = dice_metric_inv(pred_bin, gt).item()
        all_d.append(d)

        gt_plot = 1 - gt[0,0].cpu()
        pred_plot = 1 - pred_bin[0,0].cpu()

        # Visualization
        fig, axs = plt.subplots(1, 3, figsize=(12, 4))
        axs[0].imshow(img[0,0].cpu(), cmap="gray")
        axs[0].set_title("Test Image")
        axs[1].imshow(gt_plot, cmap="gray")
        axs[1].set_title("Ground Truth")
        axs[2].imshow(pred_plot, cmap="gray")
        axs[2].set_title(f"Predicted Mask, Dice {d:.4f}")
        for a in axs: a.axis("off")
        plt.show()

mean = np.mean(all_d)
std = np.std(all_d)
print(f"U-Net TEST SET Performance| mean Dice={mean:.4f} | std={std:.4f}")
````

````python
# Eval U-Net ViT
all_d = []
model_vit.eval()

with torch.no_grad():
    for i, (img, gt) in enumerate(test_loader):
        img = img.to(device)
        gt = gt.to(device)

        out = model_vit(img)
        if isinstance(out, tuple):
            out = out[0]

        pred = torch.sigmoid(out)
        pred_bin = (pred > 0.5).float()
    
        d = dice_metric_inv(pred_bin, gt).item()
        all_d.append(d)

        gt_plot = 1 - gt[0,0].cpu()
        pred_plot = 1 - pred_bin[0,0].cpu()

        # Visualization
        fig, axs = plt.subplots(1, 3, figsize=(12, 4))
        axs[0].imshow(img[0,0].cpu(), cmap="gray")
        axs[0].set_title("Test Image")
        axs[1].imshow(gt_plot, cmap="gray")
        axs[1].set_title("Ground Truth")
        axs[2].imshow(pred_plot, cmap="gray")
        axs[2].set_title(f"Predicted Mask, Dice {d:.4f}")
        for a in axs: a.axis("off")
        plt.show()

mean = np.mean(all_d)
std = np.std(all_d)
print(f"U-Net ViT TEST SET Performance| mean Dice={mean:.4f} | std={std:.4f}")
````
